```python
from matminer.utils.data import MixingEnthalpy
from pymatgen.core import Composition, Element
from matminer.featurizers.composition.alloy import WenAlloys
from matminer.featurizers.composition import ElementProperty
import itertools
```

```python
wen = WenAlloys(impute_nan=True)
mix_enthalpy_table = MixingEnthalpy(impute_nan=True)
```

```python
def calculate_wenalloy_features(composition_str, verbose=False):
    comp = Composition(composition_str)

    elements = [Element(str(e)) for e in comp.elements]
    fractions = [comp.get_atomic_fraction(e) for e in comp.elements]
    radii = [e.atomic_radius for e in elements]

    delta = wen.compute_delta(radii, fractions)
    entropy = abs(wen.compute_configuration_entropy(fractions))

    enthalpy = 0
    for (e1, e2) in itertools.combinations(elements, 2):
        c1 = comp.get_atomic_fraction(e1)
        c2 = comp.get_atomic_fraction(e2)
        h_ij = mix_enthalpy_table.get_mixing_enthalpy(e1, e2)
        if h_ij is None:
            h_ij = 0
        pair_contribution = 4 * c1 * c2 * h_ij
        enthalpy += pair_contribution
        if verbose:
            print(f"{e1}-{e2}: ΔH = {h_ij:+.2f} kJ/mol, "
                      f"weighted = {pair_contribution:+.4f}")

    if verbose:
        print(f"\nComposition: {composition_str}")
        print(f"Δ (atomic size mismatch): {delta:.4f}")
        print(f"ΔH_mix (total): {enthalpy:.4f}")
        print(f"ΔS_conf: {entropy:.4f}")
        print("---- Binary contributions ----")

    # magpie_features_dict = {}
    # if df is not None:
    #     ep = ElementProperty.from_preset("magpie")
    #     # Prepare a temporary DataFrame for this single composition
    #     temp_df = pd.DataFrame({"composition": [composition_str]})
    #     temp_df = ep.featurize_dataframe(temp_df, "composition", ignore_errors=True)
    #     magpie_features_dict = temp_df.drop(columns=["composition"]).iloc[0].to_dict()
    
    return delta, enthalpy, entropy
```

```python
from matminer.featurizers.composition.element import BandCenter
```

```python
bandObj = BandCenter(impute_nan=True)
```

```python
def calculate_band_features(composition_str, verbose=False):
    comp = Composition(composition_str)

    return bandObj.featurize(comp)[0]
```

```python
atomic_size_mismatch = []
mixing_enthalpy = []
mixing_entropy = []
formation_amor = []
toplogical_formation = []
band_sizes = []
# standard_formation = []

for _, row in df.iterrows():
    comp_str = row['Material compositions']
    trg = row['Trg']  # pull Tg value for this row
    
    # WenAlloy features
    delta, enthalpy, entropy = calculate_wenalloy_features(comp_str)
    atomic_size_mismatch.append(delta)
    mixing_enthalpy.append(enthalpy)
    mixing_entropy.append(entropy)
    
    # Miedema formation enthalpy
    deltaH, deltaH_topo = calculate_miedema_features(comp_str, trg, enthalpy)
    formation_amor.append(deltaH)
    toplogical_formation.append(deltaH_topo)
    # standard_formation.append(standard_deltaH)

    # Band featurization
    band = calculate_band_features(comp_str)
    band_sizes.append(band)
```
