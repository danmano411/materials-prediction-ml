```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/Metallic Glass Transition Data/Metallic_Glass_Forming_with_features.csv")
```

```python
# Recall that many features had 2 unique values which allow the feature to now be a binary datatype
feature_col = df.iloc[:, 3:].columns
binary_col = []

for col in feature_col:
    nunique = df[col].nunique()
    if nunique == 2:
        # print(col)
        binary_col.append(col)

binary_col
print(df['NsUnfilled_difference'])
```

```python
for col in binary_col:
    unique_values = df[col].unique()
    df[col] = df[col].map({unique_values[0]:0, unique_values[1]:1})
```

```python
#Confirm that our conversion was successful
print(df['NsUnfilled_difference'])
```

```python
# Now, create element composition features from the data found in the string id of the alloys

# Create a general function to extract composition from string
# input: A35B14.6CD50.4 => output: {'A': 35%, 'B': 14.6%, 'CD': 50.4%}
import string
import math
def composition_from_string(composition_str):
    element = ""
    amt = ""
    composition = {}
    total_assigned = 0.0

    for i, c in enumerate(composition_str):
        if c.isalpha():
            # If a new uppercase starts and there’s a previous element
            if element != "" and c.isupper():
                # Assign previous element (special case where no amt is given, 1% is assumed)
                if amt == "":
                    composition[element] = 1.0
                    total_assigned += 1.0
                else:
                    composition[element] = float(amt)
                    total_assigned += float(amt)
                element = ""
                amt = ""
            element += c
        else:
            amt += c

    # Assign the last element
    if amt == "":
        composition[element] = round(max(1.0, 100 - total_assigned), 3)
    else:
        composition[element] = float(amt)

    # Optional: if last element should take the remainder to reach 100
    if sum(composition.values()) < 100:
        last_element = list(composition.keys())[-1]
        composition[last_element] += round(100 - sum(composition.values()), 3)

    return composition
```

```python
all_elements = set()
for comp_str in df.iloc[:, 0]:
    composition = composition_from_string(comp_str)
    all_elements.update(composition.keys())

for element in all_elements:
    df[element] = df.iloc[:, 0].apply(
        lambda x: composition_from_string(x).get(element, 0)
    )

print(df.head())
```

```python
# Calculate the nuber of outliers per descriptor
from collections import Counter
outlier_indicies = []

# print(feature_col)
# print(df.iloc[:, 2:].columns)

for i, col in enumerate(feature_col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3-Q1

    outliers = df[(df[col] < (Q1-1.5*IQR)) | (df[col] > (Q3+1.5*IQR))]

    if len(outliers) != 0:
        outlier_indicies.extend(outliers.index.tolist())
c = Counter(outlier_indicies)
multiple_outliers = [val for val in set(outlier_indicies) if c[val] > 2]
multiple_outliers.sort()

# Index values where features have more than 2 property outliers
print(multiple_outliers, "\n\nTotal features with >2 outliers: ", len(multiple_outliers))
```

```python
df.drop(multiple_outliers, inplace=True)
```

```python
# Get the index of the one value with Trg < 0.3
trg_outlier = df[(df['Trg'] < 0.3)].index[0]

print("Index of the Trg outlier: ", trg_outlier)
```

```python
df.drop(trg_outlier, inplace=True)
```

```python
df.shape
```

```python
# Drop highly correlated features

# Make correlation matrix and pick unique pairs
corr_matrix = df.iloc[:, 3:].corr().abs() 
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

# Find corrlated columns with selected threshold
to_drop = [col for col in upper.columns if any(upper[col] > 0.95)]

print(df.shape)

# drop those columns
df.drop(columns=to_drop, inplace=True)

print(df.shape)
print(to_drop)
```

```python
# Highly constant threshold
threshold = 0.99

# Find the constant columns
constant_cols = [col for col in df.columns if df[col].value_counts(normalize=True).max() > threshold]

# Drop these columns
df.drop(columns=constant_cols, inplace=True)

print("Dropped columns:", constant_cols)
print(df.shape)
```

```python
df = df.reset_index(drop=True)
df.to_csv("../data/cleaned/Metallic Glass Transition Data/Metallic_Glass_Forming_with_features_CLEANED.csv", index=False)
```
