```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
feature_col = df.columns[3:]
```

```python
import math

items_per_row = 2
fig, axes = plt.subplots(math.ceil(len(feature_col)/items_per_row), items_per_row, figsize=(15, len(feature_col)*4))
axes = axes.flatten()

for i, col in enumerate(feature_col):
    skew_value = df[col].skew()
    sns.histplot(df[col], kde=True, bins=45, ax=axes[i])
    axes[i].set_title(f"Histogram of {col} (Skew: {skew_value:.2f})")
    axes[i].set_xlabel(col)
    axes[i].set_ylabel("Frequency")

plt.subplots_adjust(hspace=0.3, wspace=0.2) 
plt.show()
```

```python
# Interesting trends in unique values in each feature from looking at the data and graphing it

for i, col in enumerate(feature_col):
    print(f"Unique values of {col}: {df[col].nunique()}")
```

```python
# Consider eliminating outliers

# Visibly see any outliers
items_per_row = 2
fig, axes = plt.subplots(math.ceil(len(feature_col)/items_per_row), items_per_row, figsize=(15, len(feature_col)*4))
axes = axes.flatten()

for i, col in enumerate(feature_col):
    skew_val = df[col].skew()
    sns.boxplot(x=df[col], ax=axes[i])
    axes[i].set_title(f"Boxplot of {col} (Skew: {skew_val:.2f})")

plt.subplots_adjust(hspace=0.3, wspace=0.2) 
plt.show()
```

```python
# Calculate the nuber of outliers per descriptor
from collections import Counter
outlier_indicies = []

# print(feature_col)
# print(df.iloc[:, 2:].columns)

for i, col in enumerate(feature_col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3-Q1

    outliers = df[(df[col] < (Q1-1.5*IQR)) | (df[col] > (Q3+1.5*IQR))]

    if len(outliers) != 0:
        outlier_indicies.extend(outliers.index.tolist())
c = Counter(outlier_indicies)
multiple_outliers = [val for val in set(outlier_indicies) if c[val] > 2]
multiple_outliers.sort()

# Index values where features have more than 2 property outliers
print(multiple_outliers, "\n\nTotal features with >2 outliers: ", len(multiple_outliers))
```

```python
corr_matrix = df.iloc[:, 3:].corr()
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix,
             annot=True,
             fmt='.2f',
             cmap="coolwarm",
             cbar=True,
             mask=mask,
             vmin=-1,
             vmax=1)
plt.title("Correlation heatmap of features")
plt.show()
```

```python
trg_corr = df.iloc[:, 2:].corr()

plt.figure(figsize=(12, 8))
sns.heatmap(
    trg_corr[['Trg']].sort_values(by='Trg', ascending=False),
    fmt='.2f',
    cmap="coolwarm",
    cbar=True,
    vmin=-1,
    vmax=1,
    annot=True
)
plt.title("Correlation of descriptors with Trg")
plt.show()
```
